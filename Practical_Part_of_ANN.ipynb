```python
# Step 1: Load and preprocess dataset
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
cancer_dataset = load_breast_cancer()

print(cancer_dataset)
```

```python
# Divide data into X , y

X = cancer_dataset['data']
X
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8, random_state=42)
```

```python
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_train_scaled
```

```python
X_test_scaled = scaler.fit_transform(X_test)
X_test_scaled
```

```python
# H/w : All Binary ML Model
```

```python
# Create FNN: ANN: Feed Forward neural network
# Artificial Neural Network
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adagrad, Adam, SGD
```

image.png

```python
X.shape

X_train[0].shape
```

```python
# Create A Model

Model = Sequential([
    Dense(64, input_shape = X_train_scaled[0].shape, activation='relu'), # First Hidden layer
    Dense(64, activation='relu'), # H1
    Dense(128,activation='relu'), # H2
    Dense(32, activation='relu'), # H3
    Dense(1, activation='sigmoid')  # Output layer
])
```

```python
Model.summary()
```

```python
# Complie This Model

Model.compile(optimizer = Adagrad(learning_rate=0.001), loss = 'categorical_crossentropy', metrics=['accuracy'])
```

```python
# Train the Model

Model.fit(X_train_scaled,y_train, epochs=500, validation_split=0.2, verbose=1) #  batch_size=50
```
